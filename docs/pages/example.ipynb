```python
%%snippet-cast --tts say --order exec --highlight-color #DDDEDF --state-bg-color #DDDEDF 
                  #: typing // completing
def fib(n):   #: typing // function call / function definition 
    if n <= 1:     #: typing // expression / assignment 
        return n #: typing // expression / assignment 
    return fib(n - 1) + fib(n - 2)      #: typing // expression / return 

x = 4             #: typing // expression / assignment 
y = fib(x)        #: typing // function call / completing 
print(y)          #: typing // expression / print 
```
